```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
import os

# Lấy đường dẫn đến thư mục Downloads của người dùng
user_directory = os.path.expanduser("~")
input_file = os.path.join(user_directory, 'Downloads', 'Copy of MVR (1).xlsx')

# Đọc file
df1 = pd.read_excel(input_file)

# Giả sử dữ liệu có cột Date, Close (giá đóng cửa) và Volume (khối lượng giao dịch)
# Dữ liệu của bạn là 'df1', nơi 'Date' là cột ngày, 'Close' là giá đóng cửa và 'Volume' là khối lượng giao dịch
df1['Date'] = pd.to_datetime(df1['Date'])
df1.set_index('Date', inplace=True)

# Tính toán tỷ lệ thay đổi của giá cổ phiếu
df1['Price Change'] = df1['Close_M'].pct_change()  # Tính toán % thay đổi giá cổ phiếu

# Quy tắc Mua, Bán, Giữ:
# Mua nếu giá cổ phiếu tăng trên 5% và khối lượng giao dịch trên 100,000
# Bán nếu giá cổ phiếu giảm dưới 5% và khối lượng giao dịch dưới 100,000
df1['Action'] = np.where((df1['Price Change'] > 0.05) & (df1['Volume_M'] > 100000), 'Buy', 
                         np.where((df1['Price Change'] < -0.05) & (df1['Volume_M'] < 100000), 'Sell', 'Hold'))

# Hiển thị kết quả mô hình quyết định
print(df1[['Close_M', 'Volume_M', 'Price Change', 'Action']].tail())

# Tính toán tỷ lệ thay đổi của giá cổ phiếu (Visa)
df1['Price Change'] = df1['Close_V'].pct_change()  # Tính toán % thay đổi giá cổ phiếu

# Quy tắc Mua, Bán, Giữ:
# Mua nếu giá cổ phiếu tăng trên 5% và khối lượng giao dịch trên 100,000
# Bán nếu giá cổ phiếu giảm dưới 5% và khối lượng giao dịch dưới 100,000
df1['Action'] = np.where((df1['Price Change'] > 0.05) & (df1['Volume_V'] > 100000), 'Buy', 
                         np.where((df1['Price Change'] < -0.05) & (df1['Volume_V'] < 100000), 'Sell', 'Hold'))

# Hiển thị kết quả mô hình quyết định
print(df1[['Close_V', 'Volume_V', 'Price Change', 'Action']].tail())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
import os

# Lấy đường dẫn đến thư mục Downloads của người dùng
user_directory = os.path.expanduser("~")
input_file = os.path.join(user_directory, 'Downloads', 'Copy of MVR (1).xlsx')

# Đọc file
df1 = pd.read_excel(input_file)

# Giả sử dữ liệu có cột Date và Close_M, Close_V (giá đóng cửa của MasterCard và Visa)
df1['Date'] = pd.to_datetime(df1['Date'])
df1.set_index('Date', inplace=True)

# Dự báo cho cổ phiếu của MasterCard (Close_M)
y_mastercard = df1['Close_M']

# Áp dụng mô hình ARIMA (ví dụ ARIMA(5,1,0)) cho MasterCard
model_mastercard = ARIMA(y_mastercard, order=(5,1,0))  # Tùy chỉnh theo nhu cầu
model_fit_mastercard = model_mastercard.fit()

# Dự báo giá trong tương lai (ví dụ: 24 tháng tới)
forecast_steps = 24  # Dự báo 24 tháng tới
forecast_mastercard = model_fit_mastercard.forecast(steps=forecast_steps)

# Vẽ đồ thị kết quả cho MasterCard
plt.figure(figsize=(10,6))
plt.plot(y_mastercard, label='Historical Data (MasterCard)')
plt.plot(pd.date_range(y_mastercard.index[-1], periods=forecast_steps+1, freq='M')[1:], forecast_mastercard, label='Forecast (MasterCard)', color='red')
plt.title('MasterCard Stock Price Forecast (ARIMA)')
plt.legend()
plt.show()

# Dự báo cho cổ phiếu của Visa (Close_V)
y_visa = df1['Close_V']

# Áp dụng mô hình ARIMA (ví dụ ARIMA(5,1,0)) cho Visa
model_visa = ARIMA(y_visa, order=(5,1,0))  # Tùy chỉnh theo nhu cầu
model_fit_visa = model_visa.fit()

# Dự báo giá trong tương lai (ví dụ: 24 tháng tới)
forecast_steps = 24  # Dự báo 24 tháng tới
forecast_visa = model_fit_visa.forecast(steps=forecast_steps)

# Vẽ đồ thị kết quả cho Visa
plt.figure(figsize=(10,6))
plt.plot(y_visa, label='Historical Data (Visa)')
plt.plot(pd.date_range(y_visa.index[-1], periods=forecast_steps+1, freq='M')[1:], forecast_visa, label='Forecast (Visa)', color='blue')
plt.title('Visa Stock Price Forecast (ARIMA)')
plt.legend()
plt.show()

# Dự báo cho cổ phiếu của MasterCard (Close_M)
y_mastercard = df1['Volume_M']

# Áp dụng mô hình ARIMA (ví dụ ARIMA(5,1,0)) cho MasterCard
model_mastercard = ARIMA(y_mastercard, order=(5,1,0))  # Tùy chỉnh theo nhu cầu
model_fit_mastercard = model_mastercard.fit()

# Dự báo giá trong tương lai (ví dụ: 24 tháng tới)
forecast_steps = 24  # Dự báo 24 tháng tới
forecast_mastercard = model_fit_mastercard.forecast(steps=forecast_steps)

# Vẽ đồ thị kết quả cho MasterCard
plt.figure(figsize=(10,6))
plt.plot(y_mastercard, label='Historical Data (MasterCard)')
plt.plot(pd.date_range(y_mastercard.index[-1], periods=forecast_steps+1, freq='M')[1:], forecast_mastercard, label='Forecast (MasterCard)', color='red')
plt.title('MasterCard Stock Price Forecast (ARIMA)')
plt.legend()
plt.show()

# Dự báo cho cổ phiếu của MasterCard (Close_M)
y_mastercard = df1['Volume_V']

# Áp dụng mô hình ARIMA (ví dụ ARIMA(5,1,0)) cho MasterCard
model_mastercard = ARIMA(y_mastercard, order=(5,1,0))  # Tùy chỉnh theo nhu cầu
model_fit_mastercard = model_mastercard.fit()

# Dự báo giá trong tương lai (ví dụ: 24 tháng tới)
forecast_steps = 24  # Dự báo 24 tháng tới
forecast_mastercard = model_fit_mastercard.forecast(steps=forecast_steps)

# Vẽ đồ thị kết quả cho MasterCard
plt.figure(figsize=(10,6))
plt.plot(y_mastercard, label='Historical Data (MasterCard)')
plt.plot(pd.date_range(y_mastercard.index[-1], periods=forecast_steps+1, freq='M')[1:], forecast_mastercard, label='Forecast (MasterCard)', color='red')
plt.title('MasterCard Stock Price Forecast (ARIMA)')
plt.legend()
plt.show()
```
